```python
import pandas as pd

# 读取xls数据
# @param filepath: xls文件路径
# @param num: 记录数
# @Return tuple(文本内容数组, 标签数组)
def read_data(filepath, num=None) -> tuple[list, list]:
    data = pd.read_excel(filepath, engine='xlrd')
    if num is not None:
        data = data.iloc[:num]

    texts = data['text'].tolist()
    labels = data['label'].tolist()

    return texts, labels

import re

# 清理文本
# @param string: 文本字符串
# @Return 正则后的字符串
def clean_str(string) -> str:
    # 1. 移除所有非字母数字字符（保留空格）
    string = re.sub(r"[^A-Za-z0-9]+", " ", string)
    
    # 2. 处理常见缩写（保留缩写形式，避免拆分）
    string = re.sub(r"\'s", " 's", string)   # it's → it 's
    string = re.sub(r"\'ve", " 've", string) # you've → you 've
    string = re.sub(r"n\'t", " n't", string) # don't → do n't
    string = re.sub(r"\'re", " 're", string) # they're → they 're
    string = re.sub(r"\'d", " 'd", string)   # he'd → he 'd
    string = re.sub(r"\'ll", " 'll", string) # they'll → they 'll
    
    # 3. 移除标点符号（直接删除而非替换为空格）
    string = re.sub(r"[,!.?()]", "", string)
    
    # 4. 合并多个空格为一个
    string = re.sub(r"\s+", " ", string)
    
    # 5. 处理重复字符（如 sssss → s）
    string = re.sub(r'(.)\1{3,}', r'\1', string)
    
    # 6. 转小写并去首尾空格
    return string.strip().lower()







imdb_test = os.path.join(imdb_dir, "Test.xls")
imdb_test_texts, imdb_test_labels = read_data(imdb_test, 5)

print(imdb_test_texts)
print(imdb_test_labels)
```
